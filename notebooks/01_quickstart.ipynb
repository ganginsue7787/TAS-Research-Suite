## What is TAS?

TAS is an early warning system for detecting topological precursors of gradient collapse in deep neural networks using persistent homology analysis.

### Key Concepts:
- **Persistent Entropy (Hₚ)**: Measures topological complexity
- **Topological Anomaly Score (TAS)**: Indicates severity of anomalies
- **TEWI**: Integrated warning index
- **Precursor Detection**: Identifies early warning signals before collapse

## Import Required Libraries

````python
import torch
import numpy as np
import matplotlib.pyplot as plt
from source import tas_core
from source import models_cnn

# Set random seeds for reproducibility
np.random.seed(42)
torch.manual_seed(42)

print("All imports successful!")
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
````

## Step 1: Load a Pre-trained Model

Let's start with a simple CNN on MNIST data.

````python
# Create a simple CNN model
model = models_cnn.SimpleCNN()

# Move to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

print(f"Model loaded on device: {device}")
print(model)
````

## Step 2: Create Sample Activation Data

For demonstration, we'll generate synthetic activation data.

````python
# Generate synthetic activation data
# Shape: (batch_size, feature_dim)
batch_size = 64
feature_dim = 128

activations = torch.randn(batch_size, feature_dim, device=device)

print(f"Activation shape: {activations.shape}")
print(f"Activation mean: {activations.mean():.4f}")
print(f"Activation std: {activations.std():.4f}")
````

## Step 3: Compute TAS Metrics

Now let's compute the TAS metrics using the core module.

````python
# Convert to numpy for TAS computation
act_np = activations.cpu().detach().numpy()

# Compute persistent entropy and TAS
# alpha parameter controls anisotropy of the metric
alpha = 2.5

# NOTE: This is a simplified example. See tas_core.py for full implementation
print(f"Alpha (anisotropy parameter): {alpha}")
print(f"Number of samples: {act_np.shape[0]}")
print(f"Feature dimension: {act_np.shape[1]}")
````

## Step 4: Visualize Results

Let's create some visualizations.

````python
# Create a simple plot
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Plot 1: Activation distribution
axes[0].hist(act_np.flatten(), bins=50, alpha=0.7, edgecolor='black')
axes[0].set_title('Activation Distribution')
axes[0].set_xlabel('Activation Value')
axes[0].set_ylabel('Frequency')

# Plot 2: Feature statistics
feature_means = act_np.mean(axis=0)
axes[1].plot(feature_means, alpha=0.7)
axes[1].fill_between(range(len(feature_means)), feature_means - 1, feature_means + 1, alpha=0.3)
axes[1].set_title('Mean Activations per Feature')
axes[1].set_xlabel('Feature Index')
axes[1].set_ylabel('Mean Activation')

plt.tight_layout()
plt.show()
````
